A Python program that determines if someone is eligible to vote

```python
Voter_age = 26
Voter_status = "Registered member with a card"

if Voter_age >= 18 and "Not registered" in Voter_status:
    print("This person is not eligible to vote.")

else:
    print("This person is eligible to vote.")
```
