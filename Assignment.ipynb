List is a collection data type that is mutable.

```python
food1 = ["rice", "beans", "oatmeal", "plaintain","noodles", "egg" ]
millicent = food1 [1]
millz = food1 [3]
milly = food1 [0]
print (f"Please I need a plate of {millicent},  {millz} and {milly}.")
```

```python
fav_food = ["fried rice", "jollof rice", "beef", "moi-moi", "coleslaw"]
fav_food [2] = "chicken"
print (fav_food)
```

```python
fav_food = ["fried rice", "jollof rice", "beef", "moi-moi", "coleslaw"]
fav_food.insert (2, "turkey")
print (fav_food)
```

```python
fruits = ["paw-paw", "apple", "banana", "orange", "avocado"]
fruits.append ("mango")
print (fruits)
```

```python
fruits = ["apple", "groundnut", "orange", "avocado", "mango"]
fruits.pop (1)
print (fruits)
```

```python
car_list = ["benz", "camry","SUV", "lexus"]
car_list.extend (["porshe", "ford", "BMW"])
print (car_list)
```

Tuples store multiple items in a variable. it supports mixed data type, and is not mutable

```python
my_tup = ("notebooks", "pencils", "rulers", "textbooks")
print (f"These are the items in my bag: {my_tup}")
print (len(my_tup))
```

```python
Tuple1 = ("shoes", "handbag", "schoolbag", " belt")
Tuple2 = (1, 2, 3, 4, 5)
Tuple3 = (True, False, False)
Tuple4 = Tuple1 + Tuple2 + Tuple3
print (Tuple4)
```

Converting a tuple to a list, to mutate it.

```python
Millicent = ("shoes", "handbag", "schoolbag", " belt")
milly = list (Millicent)
milly.append ("sunglasses")
Millicent = tuple (milly)
print (Millicent)
```

```python
Udeme = ("Bible", "stick", "diary", "marker")
udy = list (Udeme)
udy [1] = "pen"
Udeme = tuple (udy)
print (Udeme)
```
