For Loop

```python
for num in range (1, 10):
    if num % 3 == 0 and num % 5 == 0:
        print ("fizzbuzz")
    elif num % 3 == 0:
        print ("fizz")
    elif num % 5 == 0:
        print ("buzz")
    else:
        print (num)
```

While Loop

```python
Total_sum = 0
num = int (input ("Enter a number: "))
while num > 0:
  Total_sum += num
  num = int (input ("Enter another number: "))
print (f"The sum of all positive number is : ", Total_sum)
```
