Phonebook Application

```python
contacts = []
try:
  with open ("contacts.txt", "r") as file:
    for line in file:
      name, phone_number_str, email, work = line.strip().split("|")
      contacts.append ({
          "name": name,
          "phone_number": phone_number_str,
          "email": email,
          "work": work
      })
except FileNotFoundError:
  contacts = []

def save_contacts ():
  with open ("contacts.txt", "w") as file:
    for contact in contacts:
      file.write (
          f"{contact['name']}|"
          f"{contact['phone_number']}|"
          f"{contact['email']}|"
          f"{contact['work']}\n"
           )

def add_contact():
  name = input ("Enter contact name:")
  for contact in contacts:
      if name.lower() == contact ["name"].lower ():
        print ("Contact already exists")
        return

  phone_number_input = input ("Enter phone number:")
  if not (phone_number_input.isdigit() and len (phone_number_input) == 11):
        print ("Invalid phone number. Please enter an 11-digit number.")
        return

  email = input ("Enter email (optional):")
  work = input ("Enter work (optional):")

  contact = {
      "name": name,
      "phone_number": phone_number_input,
      "email": email if email else None,
      "work": work if work else None
  }
  contacts.append(contact)
  save_contacts()
  print ("Contact added successfully")

def view_contacts ():
  if len (contacts) == 0:
    print ("No contacts found")
    return
  print ("\nContact List")
  sorted_contacts = sorted (contacts, key = lambda contact: contact ["name"].lower())
  current_letter = ""
  for contact in sorted_contacts:
    first_letter = contact ["name"].strip()[0].upper()
    if first_letter != current_letter:
      current_letter = first_letter
      print (f"\n{current_letter})")
    print (f" - {contact ['name']}")
    print ()

def view_contact_details ():
  name = input ("Enter contact name:")
  for contact in contacts:
    if contact ["name"].lower() == name.lower ():
      print (contact)
      print ("\nContact details:")
      print ("Name:", contact ["name"])
      print ("Phone_number:", contact ["phone_number"])
    if contact ["email"]:
      print ("Email:", contact ["email"])
    if contact ["work"]:
      print ("Work:", contact ["work"])
    return
  print ("Contact not found")

def search_contacts ():
  search_term = input ("Enter name or phone number:")
  found = False
  print ("\nMatching Contacts:")
  for contact in contacts:
    if search_term.lower() in contact ["name"].lower():
      print ("-", contact ["name"])
      found = True
    elif search_term in contact ["phone_number"]:
      print ("-", contact ["name"])
      found = True
  if not found:
    print ("No matching contact found")


def update_contact ():
  update_choice = input ("Enter contact name to update:")
  for contact in contacts:
    if update_choice.lower() == contact ["name"].lower():
      print ("1. Update Name")
      print ("2. Update Phone Number")
      print ("3. Update Email")
      print ("4. Update Work")
      choice = input ("Enter your choice:")
      if choice == "1":
        contact ["name"] = input ("Enter new name:")
        save_contacts ()
        print ("Contact successfully updated")
      elif choice == "2":
        new_phone_number_input = input ("Enter new phone number:")
        if new_phone_number_input.isdigit() and len (new_phone_number_input) == 11:
          contact ["phone_number"] = new_phone_number_input
          save_contacts ()
          print ("Contact successfully updated")
        else:
          print ("Invalid phone number.")
      elif choice == "3":
          contact ["email"] = input ("Enter new email:")
          save_contacts ()
          print ("Contact successfully updated")
      elif choice == "4":
          contact ["work"] = input ("Enter new work:")
          save_contacts ()
          print ("Contact successfully updated")
      else:
          print ("Invalid choice")
      return
  print ("Contact not found.")

def delete_contact ():
  delete_name = input ("Enter contact name to delete:")
  for contact in contacts:
    if delete_name.lower() == contact ["name"].lower():
      contacts.remove (contact)
      save_contacts ()
      print ("Contact successfully deleted")
      return
  print ("Contact not found.")


while True :
  print ("\nPHONEBOOK MENU")
  print ("1. Add contact")
  print ("2. View contacts")
  print ("3. View contact details")
  print ("4. Search contacts")
  print ("5. Update contact")
  print ("6. Delete contact")
  print ("7. Exit")
  choice = input ("Enter your choice:")
  if choice == "1":
    add_contact ()
  elif choice == "2":
    view_contacts ()
  elif choice == "3":
    view_contact_details ()
  elif choice == "4":
    search_contacts ()
  elif choice == "5":
    update_contact ()
  elif choice == "6":
    delete_contact ()
  elif choice == "7":
    print ("Goodbye")
    break
  else:
    print ("Invalid choice")
```
